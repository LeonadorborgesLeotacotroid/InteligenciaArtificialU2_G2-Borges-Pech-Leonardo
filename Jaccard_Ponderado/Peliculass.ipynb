```python
import math
todos_generos = set()
todos_actores = set()

for pelicula in peliculas.values():

    for genero in pelicula["generos"]:
        todos_generos.add(genero)

    for actor in pelicula["actores"]:
        todos_actores.add(actor)

print("Cantidad de géneros:", len(todos_generos))
print("Cantidad de actores:", len(todos_actores))
```

```python
#calcular los pesos
cantidad_peliculas = len(peliculas)

pesos_generos = {}
pesos_actores = {}

for genero in todos_generos:

    frecuencia = 0

    for pelicula in peliculas.values():

        if genero in pelicula["generos"]:
            frecuencia += 1

    pesos_generos[genero] = math.log(
        cantidad_peliculas / frecuencia
    )


for actor in todos_actores:

    frecuencia = 0

    for pelicula in peliculas.values():

        if actor in pelicula["actores"]:
            frecuencia += 1

    pesos_actores[actor] = math.log(
        cantidad_peliculas / frecuencia
    )
```

```python
def jaccard_ponderado(pelicula_a, pelicula_b):

    numerador = 0
    denominador = 0

    # Géneros
    generos_a = set(pelicula_a["generos"])
    generos_b = set(pelicula_b["generos"])

    union_generos = generos_a.union(generos_b)

    for genero in union_generos:

        peso = pesos_generos[genero]

        if genero in generos_a and genero in generos_b:
            numerador += peso

        denominador += peso


    # Actores
    actores_a = set(pelicula_a["actores"])
    actores_b = set(pelicula_b["actores"])

    union_actores = actores_a.union(actores_b)

    for actor in union_actores:

        peso = pesos_actores[actor]

        if actor in actores_a and actor in actores_b:
            numerador += peso

        denominador += peso


    if denominador == 0:
        return 0

    return numerador / denominador
```

```python
pelicula_buscada = input("Escribe una película: ")

if pelicula_buscada not in peliculas:
    print("La película no existe en el catálogo.")
else:
    print("Película seleccionada:", pelicula_buscada)
```

```python
#calcular similitudes
similitudes = []

for nombre, pelicula in peliculas.items():

    if nombre != pelicula_buscada:

        similitud = jaccard_ponderado(
            peliculas[pelicula_buscada],
            pelicula
        )

        similitudes.append((nombre, similitud))
```

```python
#ordenar
for i in range(len(similitudes)):

    for j in range(i + 1, len(similitudes)):

        if similitudes[j][1] > similitudes[i][1]:

            temporal = similitudes[i]

            similitudes[i] = similitudes[j]

            similitudes[j] = temporal
```

```python
print("\nPelícula:", pelicula_buscada)
print("\nPelículas similares:")

for i in range(5):

    nombre = similitudes[i][0]
    valor = similitudes[i][1]

    print(
        str(i + 1) + ".",
        nombre,
        "->",
        round(valor, 4)
    )
```
