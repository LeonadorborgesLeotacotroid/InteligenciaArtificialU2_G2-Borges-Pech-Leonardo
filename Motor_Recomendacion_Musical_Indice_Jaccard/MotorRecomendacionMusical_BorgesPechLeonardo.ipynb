```python
def calcular_jaccard(conjunto_a, conjunto_b):
    interseccion = len(conjunto_a.intersection(conjunto_b))
    union = len(conjunto_a.union(conjunto_b))

    if union == 0:
        return 0.0

    return interseccion / union
```

```python
cancion_actual = "Get Lucky"

atributos_actuales = catalogo_canciones[cancion_actual]

print("Canción actual:", cancion_actual)
print("Atributos:", atributos_actuales)
```

```python
similitudes = []

for cancion, atributos in catalogo_canciones.items():

    # Evitar comparar Get Lucky consigo misma
    if cancion == cancion_actual:
        continue

    similitud = calcular_jaccard(
        atributos_actuales,
        atributos
    )

    similitudes.append((cancion, similitud))
```

```python
def obtener_similitud(elemento):
    return elemento[1]

similitudes_ordenadas = sorted(
    similitudes,
    key=obtener_similitud,
    reverse=True
)

```

```python
print()
print("Top 5 recomendaciones para", cancion_actual)
print("----------------------------------------")

for i in range(5):

    cancion = similitudes_ordenadas[i][0]
    similitud = similitudes_ordenadas[i][1]

    porcentaje = similitud * 100

    print(
        i + 1,
        "-",
        cancion,
        "-",
        round(porcentaje, 2),
        "%"
    )
```
